# Ejercicio 1
Función para calcular el área de un rectángulo.
Crea una función llamada area_rectangulo() que reciba la base y la altura como parámetros. Los valores de base y altura se pedirán al usuario por teclado. Devuelve e imprime el área resultante.


```python
# Creamos la función solicitada para calcular el área de un rectángulo.
def area_rectangulo(base, altura): 
    """
    Calcula el área de un rectángulo.
    
    Parámetros: 
    base (float): La longitud de la base del rectángulo. 
    altura (float): La longitud de la altura del rectángulo.
    """
    # Hemos usado las comillas triples para documentar la función, lo que permite incluir una descripción más detallada y legible. 
    return base * altura

# Solicitud de datos al usuario para la base y la altura del rectángulo.
base = float(input("Ingrese la base del rectángulo: "))
altura = float(input("Ingrese la altura del rectángulo: "))
# Mostramos los valores ingresados por el usuario para confirmar que se han recibido correctamente. 
print(f"Base: {base}, Altura: {altura}")
# Llamada a la función para calcular el área y almacenamiento del resultado. 
area = area_rectangulo(base, altura)
# Impresión del resultado en pantalla. 
print(f"El área del rectángulo es: {area}")
```

# Ejercicio 2
Crea una función llamada area_circulo() que reciba el radio como parámetro. El radio se pedirá al usuario por teclado.
Devuelve e imprime el área del círculo.

```python
import math # Importamos la biblioteca math para usar la constante pi y otras funciones matemáticas. 
# Creamos la función solicitada para calcular el área de un circulo. 
def area_circulo(radio): 
    """
    Calcula el área de un círculo. 
    
    Parámetros: 
    radio (float): El radio del círculo. 
    Retorna: 
    float: El área del círculo calculada usando la fórmula A = π * r^2, donde A es el área y r es el radio.
    """
    pi = math.pi # Usamos la constante pi de la biblioteca math para mayor precisión. 
    return pi * radio ** 2
# Usamos el operador de exponenciación (**) para elevar el radio al cuadrado, lo que es más legible y directo que usar la función pow().
# Solicitud de datos al usuario que ingrese el radio del círculo.
radio = float(input("Ingrese el radio del círculo: "))
# Mostramos el valor ingresado por el usuario para confirmar que se ha recibido correctamente.
print(f"Radio: {radio}")
# Llamada a la función para calcular el área y almacenamiento del resultado. 
area = area_circulo(radio)
# Impresión del resultado en pantalla. 
print(f"El área del círculo es: {area}")
```

# Ejercicio 3
Crea una función llamada lee_numero() que pida al usuario un número por teclado. Usa esta función para solicitar tres números. Implementa otra función llamada mayor() que reciba tres parámetros y devuelva el mayor de los tres. Muestra el resultado.

```python
# Función para pedir un número con validación (repite el bucle mientras haya un error)
def lee_numero(mensaje):
    """Solicita un número y repite la petición si el usuario ingresa texto."""
    while True:
        try:
            # Si el usuario introduce un número, lo devuelve y sale del bucle 'while'
            return float(input(mensaje))
        except ValueError:
            # Si introduce texto, muestra el mensaje y el 'while' vuelve a empezar
            print("¡Error! Por favor, ingrese un número válido.")


# Función para calcular el mayor de tres números
def mayor(n1, n2, n3):
    """Devuelve el número más grande entre tres valores."""
    return max(n1, n2, n3)


# --- Bloque principal ---

# Pedimos los tres números utilizando la función con 'while'
num1 = lee_numero("Ingrese el primer número: ")
num2 = lee_numero("Ingrese el segundo número: ")
num3 = lee_numero("Ingrese el tercer número: ")

# Confirmamos los números recibidos
print(f"\nLos números ingresados son: {num1}, {num2} y {num3}")

# Calculamos y mostramos el mayor
numero_mayor = mayor(num1, num2, num3)
print(f"El número mayor es: {numero_mayor}")
```

# Ejercicio 4
Crea una función llamada imc() que reciba el peso (kg) y la altura (m, con dos decimales). Calcula el índice de masa corporal: IMC = peso / (altura * altura).
Clasifica el resultado según la tabla:
Bajo peso: < 18.50
Normal: [18.50, 25.00]
Sobrepeso: ≥ 25.00
Obesidad: ≥ 30.00
Muestra el resultado con su clasificación.

```python
# Creamos la función solicitada para calcular el indice de masa corporal (IMC).
def imc(peso,altura):
    """
    Calcula el índice de masa corporal (IMC):
    IMC = peso / altura^2
    Devuelve el valor del IMC y la categoría correspondiente según los rangos establecidos por la Organización Mundial de la Salud (OMS).
    Parámetros: 
    Peso (float): El peso de la persona en kilogramos. 
    Altura (float): La altura de la persona en metros. 
    Retorna: 
    Tuple: El valor de IMC y la categoría correspondiente.
    """
    # Calculamos el IMC usando la fórmula proporcionada.
    imc_valor = peso / (altura ** 2)

    # Determinamos la categoría del IMC según los rangos establecidos por la OMS.
    if imc_valor < 18.5:
        categoria = "Bajo peso"
    elif 18.5 <= imc_valor <= 25:
        categoria = "Normal"
    elif 25 <= imc_valor < 30:
        categoria = "Sobrepeso"
    else:
        categoria = "Obesidad"

    return imc_valor, categoria

# Solicitud de datos al usuario para el peso y la altura.
peso = float(input("Ingrese su peso en kilogramos: "))
altura = float(input("Ingrese su altura en metros: "))
imc_valor, categoria = imc(peso, altura)
# Mostramos los valores ingresados por el usuario para confirmar que se han recibido correctamente.
print(f"Peso: {peso} kg, Altura: {altura} m")
print(f"IMC: {imc_valor:.2f}, Categoría: {categoria}")
```

# Ejercicio 5
Reestructura el programa del ejercicio 9 del Tema 2 dividiéndolo al máximo en funciones, por ejemplo: pedir_contraseña(), validar_login(), etc.

Programa original:

Paso 1: Solicita al usuario una contraseña y guárdala en una variable.
Paso 2: Pide al usuario que introduzca la contraseña para loguearse.
Si la contraseña es correcta, muestra "Login correcto".
Si es incorrecta, muestra "Login fallido. Intento X", incrementando un contador hasta 5.
Al llegar a 5 intentos fallidos, reinicia el programa al Paso 1.

```python
# Creamos la función solicitada para pedir una contraseña para login.
def pedir_contraseña():
    """ Solicita al usuario que ingrese una contraseña y la confirma. """
    return input("Ingrese su contraseña: ")

def validar_contraseña(contraseña):
    """ Gestiona el inicio de sesión con un máximo de 5 intentos. 
    Retorna True si la validación es exitosa, False si se agotan los intentos. """
    intentos = 0
    while intentos < 5:
        contraseña_ingresada = pedir_contraseña()
        if contraseña_ingresada == contraseña:
            print("Login correcto.")
            return True
        else:
            intentos += 1
            print(f"Login fallido. Intento {intentos}.")
    return False

def ejecutar_login():
    """ Coordina el flujo principal del programa de inicio de sesión reiniciando el mismo si se agotan los intentos. """
    while True:
        # Solicitar y guardar la contraseña inicial.
        clave_guardada = pedir_contraseña()
        # Paso 1: Solicitar al usuario que ingrese una contraseña.
        # Paso 2: Proceso de validación. Retorna True si acierta, False si se agotan los intentos.
        exito = validar_contraseña(clave_guardada)
        if exito:
            break # Salir del bucle si la validación es exitosa.
        else:
            print("Se ha agotado el número de intentos. Reiniciando el proceso de inicio de sesión...\n")
ejecutar_login() # Llamada a la función principal para iniciar el flujo del programa.
```

# Ejercicio 6 
Crea una función llamada intermedio() que reciba dos números y devuelva su punto intermedio.
Los números se deben pedir al usuario por teclado.
Muestra el resultado en pantalla.
Nota: El punto intermedio se calcula como la suma de los dos números dividida entre 2.

```python
# Creamos la función solicitada para recibir dos números y devolver su punto intermedio.
def intermedio(num1,num2):
    """
    Calcula el punto intermedio entre dos números.
    
    Parámetros:
    num1 (float): El primer número.
    num2 (float): El segundo número.
    Retorna:
    float: El punto intermedio entre los dos números.
    """
    return (num1 + num2) / 2

# Solicitud de datos al usuario por el teclado para los dos números.
numero1 = float(input("Ingrese el primer número: "))
numero2 = float(input("Ingrese el segundo número: "))
# Las variables numero1 y numero2 podrian llamarse num1 y num2, 
# pero se ha optado por no duplicar nombres de variables para evitar 
# confusiones con los parámetros de la función intermedio. 
# Pese a ser posible sintacticamente, no es recomendable usar el mismo nombre para variables locales y parámetros de funciones,
# ya que puede llevar a errores de interpretación y dificultar la lectura del código. 
# Por ello, se ha decidido mantener nombres distintos para las variables que almacenan 
# los valores ingresados por el usuario y los parámetros de la función.

# Mostramos los valores ingresados por el usuario para confirmar que se han recibido correctamente.
print(f"Número 1: {numero1}, Número 2: {numero2}")

# Llamada a la función para calcular el punto intermedio y almacenamiento del resultado. 
punto_medio = intermedio(numero1, numero2)
print(f"El punto intermedio entre {numero1} y {numero2} es: {punto_medio}")
```

# Ejercicio 7 
Crea una función llamada recortar() que reciba tres parámetros: número, límite inferior y límite superior.
Si el número es menor que el límite inferior, devolverá el límite inferior.
Si es mayor que el límite superior, devolverá el límite superior. Si está dentro del rango, se devuelve sin cambios.


```python
# Creamos la función solicitada para recibir los tres parámetros que aparecen en el enunciado.
def recortar(numero, limite_inferior, limite_superior):
    """
    Recorta un número dentro de un rango definido por un mínimo y un máximo.
    
    Parámetros:
    numero (float): El número a recortar.
    limite_inferior (float): El valor mínimo del rango.
    limite_superior (float): El valor máximo del rango.
    
    Retorna: 
    El número recortado si está fuera del rango, o el mismo número si está dentro del rango.
    """
    if numero < limite_inferior:
        return limite_inferior
    elif numero > limite_superior: 
        return limite_superior
    else:
        return numero

# Solicitud de datos al usuario por el teclado para los tres parámetros.
numero = float(input("Ingrese un número: "))
limite_inferior = float(input("Ingrese el límite inferior: "))
limite_superior = float(input("Ingrese el límite superior: "))

# Mostramos los valores ingresados por el usuario para confirmar que se han recibido correctamente. 
print(f"Número: {numero} | Límite inferior: {limite_inferior} | Límite superior: {limite_superior}")

# Llamada a la función para recortar el número y almacenamiento del resultado.
numero_recortado = recortar(numero, limite_inferior, limite_superior)
print(f"El número resultante después de recortar es: {numero_recortado}")
```

# Ejercicio 8
Crea una función llamada separar() que reciba una lista de enteros.
Debe devolver dos listas ordenadas: una con los números pares y otra con los impares.

```python
# Creamos la función solicitada para recibir una lista de números enteros y devolver dos listas ordenadas.
#  Una con los números pares y otra con los números impares.
def separar(lista_numeros):
    """
    Separa una lista de números enteros en dos listas: una con los números pares y otra con los números impares.

    Parámetros:
    lista_numeros (list): Una lista de números enteros.

    Retorna:
    tuple: Dos listas, la primera con los números pares y la segunda con los números impares.
    """
    pares = []
    impares = []
    for n in lista_numeros:
        if n % 2 == 0:
            pares.append(n)
        else:
            impares.append(n)
    return sorted(pares), sorted(impares)

# Solicitud de datos al usuario por el teclado para la lista de números enteros.
entrada = input("Ingrese una lista de números enteros separados por comas: ")
lista_numeros = [int(x) for x in entrada.split(",")]

# Mostramos los valores ingresados por el usuario para confirmar que se han recibido correctamente. 
print(f"Lista de números ingresada: {lista_numeros}")

# Llamada a la función para separar los números y almacenamiento del resultado.
lista_pares, lista_impares = separar(lista_numeros)
print(f"Números pares ordenados: {lista_pares}")
print(f"Números impares ordenados: {lista_impares}")
```
